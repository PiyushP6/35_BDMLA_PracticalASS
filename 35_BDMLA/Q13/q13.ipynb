```python
text = "  HELLO, World!! This is @ Python # Programming.  "
```

```python
text_replace = text.lower()
print("Lowercase Text:", text_replace)
```

```python
for char in [",", "!", "@", "#", "."]:
    text_replace = text_replace.replace(char, "")
print("Text after removing special characters:", text_replace)
```

```python
text_s = text_replace.strip()
print("Text after removing leading and trailing spaces:", text_s)
```

```python
text_j = " ".join(text_s.split())
print("Cleaned Text:", text_j)
```

```python
tokens = text_j.split()

print("Cleaned Text:", text_j)
print("Word Tokens:", tokens)
```
