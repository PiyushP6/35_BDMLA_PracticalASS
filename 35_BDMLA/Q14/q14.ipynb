```python
from collections import Counter

text = """
Python is a popular programming language.
Python is used in data science and machine learning.
Data science is useful and Python is easy to learn.
Machine learning is a part of data science.
"""

text = text.lower()
```

```python
for char in [".", ",", "!", "?", ";", ":"]:
    text = text.replace(char, "")
```

```python
text = " ".join(text.split())
print("Cleaned Text:", text)
```

```python
words = text.split()
print("Word Tokens:", words)
```

```python
stopwords = ["the", "is", "in", "and", "a", "to", "of", "on", "for"]

filtered_words = [word for word in words if word not in stopwords]

word_freq = Counter(filtered_words)

print("Cleaned Words:", filtered_words)
print("Top 5 Most Frequent Keywords:")

for word, frequency in word_freq.most_common(5):
    print(word, ":", frequency)
```
