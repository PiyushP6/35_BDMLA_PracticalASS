```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Piyush", "Om", "Smit", "prem", "jemish", "Rahul", "Riya", "Jay"],
    "Age": [21, 22, None, 23, 21, None, 24, None],
    "City": ["Surat", "Mumbai", "Ahmedabad", "Vadodara", "Surat", "Ahmedabad", "Mumbai", "Surat"],
    "Salary": [25000, 30000, 22000, None, 25000, 22000, 40000, 32000]
})

print(df)
```

```python
print("Missing Values:")
print(df.isna().sum())

print("\nAfter dropna():")
print(df.dropna())
```

```python
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())
df_mean["Salary"] = df_mean["Salary"].fillna(df_mean["Salary"].mean())

print("After Filling Missing Values with Mean:")
print(df_mean)

df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())
df_median["Salary"] = df_median["Salary"].fillna(df_median["Salary"].median())

print("\nAfter Filling Missing Values with Median:")
print(df_median)
```

```python
df_clean = df_mean.drop_duplicates()

print("After Removing Duplicates:")
print(df_clean)
```

```python
df_clean = df_clean.rename(columns={
    "Name": "Student_Name",
    "Age": "Student_Age",
    "City": "Student_City",
    "Salary": "Monthly_Salary"
})

print("After Renaming Columns:")
print(df_clean)
```

```python
print("Final Missing Values:")
print(df_clean.isna().sum())
```
