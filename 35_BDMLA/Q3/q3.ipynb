```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(24)
n_rows = 100_000

df = pd.DataFrame({
    'User_ID': np.random.randint(1000, 9999, size=n_rows, dtype='int64'),
    'Age': np.random.randint(18, 70, size=n_rows, dtype='int64'),
    'Account_Balance': np.random.uniform(500.0, 50000.0, size=n_rows),
    'Subscription_Type': np.random.choice(['Basic', 'Standard', 'Premium', 'VIP'], size=n_rows),
    'Active_Status': np.random.choice(['Active', 'Inactive'], size=n_rows)
})

print(df.head())
```

```python
print("Original Data Types:")
print(df.dtypes)
```

```python
before = df.memory_usage(deep=True).sum() / 1024**2

print("Memory Usage Before Optimization:")
print(f"{before:.2f} MB")
```

```python
df_c = df.copy()

df_c["User_ID"] = df_c["User_ID"].astype("int32")
df_c["Age"] = pd.to_numeric(df_c["Age"], downcast='integer')
df_c["Account_Balance"] = df_c["Account_Balance"].astype("float32")

df_c["Subscription_Type"] = df_c["Subscription_Type"].astype("category")
df_c["Active_Status"] = df_c["Active_Status"].astype("category")
```

```python
print("After Optimization Data Types:")
print(df_c.dtypes)
```

```python
after = df_c.memory_usage(deep=True).sum() / 1024**2
print("Memory Usage After Optimization:")
print(f"{after:.2f} MB")
```

```python
reduction = before - after
percentage_reduction = (reduction / before) * 100
print("Memory Usage Reduction:")
print(f"{reduction:.2f} MB")
print(f"Percentage Reduction: {percentage_reduction:.2f}%")
```
