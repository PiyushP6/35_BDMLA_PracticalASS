```python
import pandas as pd
import numpy as np
```

```python
csv_filename = "employee.csv"
records = 100000

df_temp = pd.DataFrame({
    'Employee_ID': range(1001, 1001 + records),
    'Salary': np.random.uniform(15000.0, 80000.0, size=records).round(2),
    'Department': np.random.choice(
        ['IT', 'HR', 'Finance', 'Marketing'], size=records
    )
})

df_temp.to_csv(csv_filename, index=False)

print(f"Created '{csv_filename}' with {records:,} records.")
print(df_temp.head())
```

```python
chunk_size = 10000
total_records = 0
running_sum = 0.0
chunk_number = 1

for chunk in pd.read_csv(csv_filename, chunksize=chunk_size):

    chunk_count = len(chunk)
    chunk_sum = chunk['Salary'].sum()

    total_records += chunk_count
    running_sum += chunk_sum

    print(
        f"Chunk {chunk_number}: Processed {chunk_count} rows | "
        f"Chunk Sum: Rs. {chunk_sum:,.2f} | "
        f"Running Total: Rs. {running_sum:,.2f}"
    )

    chunk_number += 1
```

```python
overall_average = running_sum / total_records 

print("FINAL AGGREGATION METRICS :")
print(f"Total Records Processed : {total_records:,}")
print(f"Total Cumulative Salary : Rs. {running_sum:,.2f}")
print(f"Overall Average Salary  : Rs. {overall_average:,.2f}")
```
