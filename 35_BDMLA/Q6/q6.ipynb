```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.appName("DS505_Transformations").master("local[*]").getOrCreate()

data = [
    (1, "Piyush","IT", 55000, "Surat"),
    (2, "Om", "HR", 75000, "Surat"),
    (3, "Smit", "IT", 80000, "Mumbai"),
    (4, "Prem", "Finance", 65000, "Surat"),
    (5, "Jemish", "HR", 90000, "Ahmedabad")
]
columns = ["ID", "Name", "Dept", "Salary", "City"]
df = spark.createDataFrame(data, columns)

df.show()
```

```python
print("Selecting 'Name' and 'Salary' columns:")
df.select("Name", "Salary").show()
```

```python
print("Department-wise Salary Filter (Dept = 'HR' and Salary > 75000):")
df.filter((col("Dept") == "HR") & (col("Salary") > 75000)).show()
```

```python
print("Add Computed Column (Bonus = 10% of Salary):")
df_with_bonus = df.withColumn("Bonus", col("Salary") * 0.10)
df_with_bonus.show()
```

```python
print("Renaming Column 'Salary' to 'Base_Salary':")
df_renamed = df_with_bonus.withColumnRenamed("Salary", "Base_Salary")
df_renamed.show()
```
