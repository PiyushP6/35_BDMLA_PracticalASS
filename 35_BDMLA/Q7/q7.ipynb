```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import count, avg, max, sum

spark = SparkSession.builder.appName("GroupedAggregations").master("local[*]").getOrCreate()

data = [
    (1, "Piyush", "IT", 50000),
    (2, "Om", "HR", 40000),
    (3, "Smit", "IT", 60000),
    (4, "Prem", "HR", 45000),
    (5, "Jemish", "Sales", 35000),
    (6, "Vansh", "Sales", 55000)
]

columns = ["EmployeeID","Name" ,"Department", "Salary"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
result = df.groupBy("Department").agg(
    count("*").alias("Employee_Count"),
    avg("Salary").alias("Average_Salary"),
    max("Salary").alias("Maximum_Salary"),
    sum("Salary").alias("Total_Salary")
)

result = result.orderBy("Total_Salary", ascending=False)

result.show()
```

```python
print("Total Departments:", result.count())

spark.stop()
```
