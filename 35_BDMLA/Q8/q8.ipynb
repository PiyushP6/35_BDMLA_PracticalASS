```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("ParquetStorage").master("local[*]").getOrCreate()

data = [
    (1, "Piyush", "IT", 50000),
    (2, "Vansh", "HR", 40000),
    (3, "Om", "Sales", 35000),
    (4, "jemish", "IT", 60000)
]

columns = ["ID", "Name", "Department", "Salary"]

df = spark.createDataFrame(data, columns)

print("Original DataFrame:")
df.show()

print("Original Schema:")
df.printSchema()
```

```python
path = "employee_data.parquet"

df.write.mode("overwrite").parquet(path)

df_parquet = spark.read.parquet(path)

print("Data Loaded from Parquet:")
df_parquet.show()

print("Loaded Schema:")
df_parquet.printSchema()

print("Original Row Count:", df.count())
print("Parquet Row Count:", df_parquet.count())

print("Schema Preserved:", df.schema == df_parquet.schema)

spark.stop()
```
